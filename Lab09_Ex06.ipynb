## คำถาม

1. Member constant คืออะไร และมีคุณสมบัติอย่างไร


ตอบ Member constant คือค่าคงที่ที่ถูกกำหนดภายในคลาสหรือโครงสร้างใน C# ซึ่งไม่สามารถเปลี่ยนแปลงได้หลังจากกำหนด และสามารถใช้ในการเก็บข้อมูลที่ไม่เปลี่ยนแปลงภายในคลาส เช่น ค่าของ Pi ในตัวอย่างข้างต้น

2. เราสามารถประกาศ member constant ในภาษา C# ได้อย่างไร


ตอบ การประกาศ member constant ในภาษา C# ทำได้โดยการใช้คำสงวน const โดยสามารถกำหนดให้เป็น static หรือ instance ขึ้นอยู่กับว่าต้องการให้ค่าคงที่นั้นเป็นค่าคงที่ของคลาสหรือออบเจ็กต์

3. member constant สามารถสร้างเป็นชนิดข้อมูลใดได้บ้าง


ตอบ ใน C# member constant สามารถประกาศเป็นหลายประเภทได้ เช่น int, double, string, char, bool, decimal, DateTime เป็นต้น ซึ่งสามารถนำมาใช้เก็บค่าคงที่ที่ไม่สามารถเปลี่ยนแปลงได้ตลอดการทำงานของโปรแกรม

4. เราสามารถเข้าถึง member constant ได้อย่างไร


ตอบ Static member constant: สามารถเข้าถึงได้โดยใช้ชื่อคลาสตามด้วยชื่อ member constant เช่น ClassName.ConstantName

Instance member constant: สามารถเข้าถึงได้ผ่านออบเจ็กต์ของคลาสนั้นๆ เช่น objectName.ConstantName (แต่มักใช้ในคลาสโดยตรงในฟังก์ชันต่างๆ)

5. member constant เป็น static โดยปริยายหรือไม่


ตอบ member constant ที่ประกาศด้วยคำสงวน const จะเป็น static โดยปริยาย ซึ่งหมายความว่าค่าคงที่นั้นสามารถเข้าถึงได้จากชื่อคลาส โดยไม่จำเป็นต้องสร้างออบเจ็กต์ของคลาส

instance member constant ยังสามารถประกาศได้ แต่โดยทั่วไปจะใช้ static สำหรับ member constant ในการเข้าถึงผ่านคลาส

6. มีข้อจำกัดอะไรบ้างในการใช้ member constant 


ตอบ ต้องกำหนดค่าในขณะประกาศ ไม่สามารถเปลี่ยนแปลงค่าได้หลังจากนั้น

ไม่สามารถใช้ค่าจากตัวแปรหรือฟังก์ชันที่ถูกคำนวณในระหว่างการทำงาน

ไม่สามารถใช้กับชนิดข้อมูลที่เป็น mutable types เช่น List หรือ Dictionary

ไม่สามารถใช้เป็น ref หรือ out parameters ได้

ไม่สามารถใช้กับประเภทเชิงซ้อน (complex types) ที่สามารถเปลี่ยนแปลงภายในได้

7. member constant เหมาะกับการใช้งานแบบไหน


ตอบ การใช้ member constant จึงเหมาะสำหรับการเก็บค่าคงที่ที่ไม่เปลี่ยนแปลงและสามารถรู้ค่าตั้งแต่ตอนคอมไพล์

8. จงตัวอย่าง Member Constant ในไลบรารี่ของภาษา C#

ตอบ ในไลบรารี่มาตรฐานของ C# มีการใช้ member constant หลายๆ ตัวอย่าง เช่น Math.PI, Int32.MaxValue, DateTime.MaxValue, และ Environment.NewLine ซึ่งทั้งหมดเป็นค่าคงที่ที่มีการประกาศในคลาสต่างๆ และสามารถนำไปใช้ในโปรแกรมของคุณได้โดยไม่ต้องสร้างออบเจ็กต์ใหม่ เหมาะสำหรับการใช้งานค่าคงที่ที่ไม่เปลี่ยนแปลง
