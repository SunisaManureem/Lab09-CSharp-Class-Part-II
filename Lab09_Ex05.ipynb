## คำถาม

1. Static Method คืออะไร


ตอบ Static Method คือเมธอดที่เกี่ยวข้องกับคลาสเอง ไม่เกี่ยวข้องกับอินสแตนซ์ของคลาส และสามารถเรียกใช้ได้โดยไม่ต้องสร้างออบเจ็กต์ของคลาสนั้น

2. ประกาศ static method ในภาษา C# อย่างไร


ตอบ การประกาศ static method ในภาษา C# ใช้คีย์เวิร์ด static ก่อนประเภทของค่าผลลัพธ์ของเมธอด (return type) ตามด้วยชื่อเมธอดและพารามิเตอร์ (ถ้ามี) เช่นเดียวกับเมธอดปกติ แต่ต้องระบุว่าเป็น static เสมอ

3. เรียกใช้ static method อย่างไร


ตอบ การเรียกใช้ static method ในภาษา C# สามารถทำได้โดยใช้ชื่อของ คลาส ตามด้วย ชื่อของเมธอด โดยไม่ต้องสร้างออบเจ็กต์ของคลาสนั้นๆ

4. static method สามารถเข้าถึง instance members (fields, methods) ได้หรือไม่


ตอบ Static method ไม่สามารถเข้าถึง instance members (เช่น fields หรือ instance methods) ได้โดยตรง เพราะ static method นั้นถูกเชื่อมโยงกับคลาสเอง ไม่ใช่ออบเจ็กต์ของคลาส ดังนั้นมันจึงไม่สามารถเข้าถึงข้อมูลหรือฟังก์ชันที่เป็นของ instance (ที่เป็นออบเจ็กต์ของคลาส) ได้

5. static method เหมาะกับการใช้งานแบบไหน


ตอบ Static method เหมาะกับการใช้งานในกรณีที่คุณต้องการทำงานที่ไม่ขึ้นกับสถานะของออบเจ็กต์ (instance) และสามารถใช้ร่วมกันได้ในระดับคลาส โดยทั่วไปแล้วจะใช้ในกรณีที่ไม่จำเป็นต้องมีข้อมูลเฉพาะตัวของออบเจ็กต์ในการทำงาน

6. ข้อดีของการใช้ static method คืออะไรบ้าง


ตอบ ไม่ต้องสร้างออบเจ็กต์

ใช้งานง่ายในคลาสเดียวกัน

ประหยัดหน่วยความจำและเพิ่มประสิทธิภาพ

สะดวกสำหรับการทำงานกับค่าคงที่และข้อมูลที่ใช้ร่วมกัน

ช่วยให้โค้ดมีความกระชับและเข้าใจง่าย

เหมาะสำหรับการทำงานกับฟังก์ชันที่ไม่ขึ้นกับสถานะของออบเจ็กต์

7. ข้อเสียของการใช้ static method คืออะไรบ้าง

8. จงยกตัวอย่าง static method ที่ใช้บ่อยใน Library มาตรฐานของภาษา C# พร้อมทั้งยกตัวอย่างการใช้งาน
