## คำถาม

1. ในภาษา C# นั้น Instance members และ static members  แตกต่างกันอย่างไร

ตอบ nstance Members ใช้สำหรับข้อมูลที่ต้องการให้แยกกันระหว่างอ็อบเจ็กต์แต่ละตัว เช่น ชื่อ อายุ หรือสถานะของอ็อบเจ็กต์

Static Members ใช้สำหรับข้อมูลที่ไม่เปลี่ยนแปลงหรือใช้ร่วมกันในทุกอ็อบเจ็กต์ของคลาสนั้น ๆ เช่น ค่าคงที่ที่ใช้ร่วมกัน ฟังก์ชันที่ไม่ต้องใช้สถานะของอ็อบเจ็กต์

2. เข้าถึง Instance members และ static members มีหลักการที่แตกต่างกันอย่างไร

ตอบ Instance Members: ต้องใช้ตัวแปรอ็อบเจ็กต์เพื่อเข้าถึง และมีค่าที่แยกกันสำหรับแต่ละอ็อบเจ็กต์ของคลาส

Static Members: สามารถเข้าถึงได้โดยตรงผ่านชื่อคลาสโดยไม่ต้องสร้างอ็อบเจ็กต์ และมีค่าเดียวกันสำหรับทุกอ็อบเจ็กต์ที่สร้างจากคลาสนั้น

3. Instance members และ static members เหมาะกับการใช้งานแบบไหน

ตอบ Instance members: ใช้เมื่อข้อมูลหรือฟังก์ชันเกี่ยวข้องกับแต่ละออบเจ็กต์ (อินสแตนซ์) ที่แตกต่างกัน

Static members: ใช้เมื่อข้อมูลหรือฟังก์ชันไม่เกี่ยวข้องกับออบเจ็กต์ใดๆ แต่ใช้ร่วมกันในระดับคลาส

4. Instance members และ static members ใช้หน่วยความจำต่างกันอย่างไร

ตอบ Instance members จะใช้หน่วยความจำแยกกันสำหรับแต่ละออบเจ็กต์

Static members จะใช้หน่วยความจำเดียวกันในระดับคลาส โดยไม่เกี่ยวข้องกับจำนวนออบเจ็กต์

5. Instance members และ static members มีอายุการใช้งานต่างกันอย่างไร

ตอบ Instance members จะมีอายุการใช้งานตามอายุของออบเจ็กต์ที่เป็นเจ้าของ (เมื่อออบเจ็กต์ถูกทำลาย, instance members จะถูกทำลาย)

Static members จะมีอายุการใช้งานตลอดระยะเวลาโปรแกรมทำงานหรือจนกว่าจะสิ้นสุดการทำงานของแอปพลิเคชัน

6. Instance members และ static members สามารถใช้งานร่วมกันได้หรือไม่

ตอบ Instance members และ static members สามารถใช้งานร่วมกันได้ แต่มีข้อจำกัดและวิธีการเข้าถึงที่แตกต่างกัน
